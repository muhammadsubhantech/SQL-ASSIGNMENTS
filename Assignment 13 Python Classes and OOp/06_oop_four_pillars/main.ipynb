### Q1. Student Result Manager | Beginner

```python
class Student:
    def __init__(self, name, roll_no):
        self.__name = name
        self.__roll_no = roll_no
        self.__marks = []

    def set_marks(self, m1, m2, m3):
        if m1 < 0 or m1 > 100 or m2 < 0 or m2 > 100 or m3 < 0 or m3 > 100:
            print("Invalid marks. Marks must be between 0 and 100.")
            return
        self.__marks = [m1, m2, m3]

    def get_average(self):
        if len(self.__marks) == 0:
            return 0
        total = self.__marks[0] + self.__marks[1] + self.__marks[2]
        return total / 3

    def display_result(self):
        avg = self.get_average()
        if avg >= 90:
            grade = "A"
        elif avg >= 80:
            grade = "B"
        elif avg >= 70:
            grade = "C"
        elif avg >= 60:
            grade = "D"
        else:
            grade = "F"
        print("Name:", self.__name)
        print("Roll No:", self.__roll_no)
        print("Marks:", self.__marks)
        print("Average:", round(avg, 2))
        print("Grade:", grade)


s1 = Student("Alice", 101)
s1.set_marks(85, 90, 78)
s1.display_result()
```

### Q2. Secure Bank Account | Intermediate

```python
class BankAccount:
    def __init__(self, account_holder):
        self.__account_holder = account_holder
        self.__balance = 0

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
            return
        self.__balance = self.__balance + amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdraw amount must be positive.")
            return
        if amount > self.__balance:
            print("Insufficient funds. Available balance:", self.__balance)
            return
        self.__balance = self.__balance - amount
        print("Withdrawn:", amount)

    def get_balance(self):
        return self.__balance

    def display(self):
        print("Account Holder:", self.__account_holder)
        print("Balance:", self.__balance)


acc = BankAccount("John")
acc.deposit(1000)
acc.withdraw(300)
print("Current Balance:", acc.get_balance())
acc.withdraw(800)
acc.display()
```

## Pillar 2: Inheritance

### Q3. Employee Hierarchy | Beginner

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


class Manager(Employee):
    def __init__(self, name, salary, bonus):
        Employee.__init__(self, name, salary)
        self.bonus = bonus

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Bonus:", self.bonus)
        print("Total:", self.salary + self.bonus)


class Developer(Employee):
    def __init__(self, name, salary, language):
        Employee.__init__(self, name, salary)
        self.language = language

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Language:", self.language)


m = Manager("Alice", 80000, 10000)
m.display()
print("---")
d = Developer("Bob", 70000, "Python")
d.display()
```

### Q4. Vehicle Rental System | Intermediate

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        self.is_rented = False

    def rent(self):
        if self.is_rented:
            print(self.brand, self.model, "is already rented.")
        else:
            self.is_rented = True
            print(self.brand, self.model, "has been rented.")

    def return_vehicle(self):
        if not self.is_rented:
            print(self.brand, self.model, "is not rented.")
        else:
            self.is_rented = False
            print(self.brand, self.model, "has been returned.")


class Car(Vehicle):
    def __init__(self, brand, model, seats):
        Vehicle.__init__(self, brand, model)
        self.seats = seats

    def rent(self):
        if self.is_rented:
            print("Car", self.brand, self.model, "is already rented.")
        else:
            self.is_rented = True
            print("Car", self.brand, self.model, "with", self.seats, "seats has been rented.")


class Bike(Vehicle):
    def __init__(self, brand, model, bike_type):
        Vehicle.__init__(self, brand, model)
        self.bike_type = bike_type

    def rent(self):
        if self.is_rented:
            print("Bike", self.brand, self.model, "is already rented.")
        else:
            self.is_rented = True
            print("Bike", self.brand, self.model, "(", self.bike_type, ") has been rented.")


c = Car("Toyota", "Corolla", 5)
c.rent()
c.rent()
c.return_vehicle()
print("---")
b = Bike("Honda", "CG125", "Standard")
b.rent()
b.return_vehicle()
```

### Q5. Notification Dispatcher | Beginner

```python
class EmailNotifier:
    def send(self, message):
        print("Sending Email:", message)


class SMSNotifier:
    def send(self, message):
        print("Sending SMS:", message)


class PushNotifier:
    def send(self, message):
        print("Sending Push Notification:", message)


def dispatch_notification(notifier, message):
    notifier.send(message)


email = EmailNotifier()
sms = SMSNotifier()
push = PushNotifier()

dispatch_notification(email, "Your order has been shipped.")
dispatch_notification(sms, "OTP is 123456")
dispatch_notification(push, "You have a new message.")
```

### Q6. Checkout Discount Calculator | Intermediate

```python
class Discount:
    def apply(self, amount):
        return amount


class PercentageDiscount(Discount):
    def __init__(self, percent):
        self.percent = percent

    def apply(self, amount):
        discount = amount * self.percent / 100
        return amount - discount


class FixedDiscount(Discount):
    def __init__(self, fixed_amount):
        self.fixed_amount = fixed_amount

    def apply(self, amount):
        if amount > self.fixed_amount:
            return amount - self.fixed_amount
        return 0


class BuyOneGetOne(Discount):
    def apply(self, amount):
        return amount / 2


def checkout(amount, discount_obj):
    final = discount_obj.apply(amount)
    print("Original:", amount, "Final after discount:", final)


checkout(1000, PercentageDiscount(10))
checkout(1000, FixedDiscount(150))
checkout(1000, BuyOneGetOne())
```

## Pillar 4: Abstraction

### Q7. Shape Calculator | Beginner

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


c = Circle(5)
print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())

r = Rectangle(4, 6)
print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())
```

### Q8. Payment Gateway | Intermediate

```python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass


class CreditCardPayment(PaymentProcessor):
    def __init__(self, card_number):
        self.card_number = card_number

    def process_payment(self, amount):
        print("Processing Credit Card payment of", amount, "using card ending with", self.card_number[-4:])
        print("Payment Successful via Credit Card")


class PayPalPayment(PaymentProcessor):
    def __init__(self, email):
        self.email = email

    def process_payment(self, amount):
        print("Processing PayPal payment of", amount, "for account", self.email)
        print("Payment Successful via PayPal")


class UPIPayment(PaymentProcessor):
    def __init__(self, upi_id):
        self.upi_id = upi_id

    def process_payment(self, amount):
        print("Processing UPI payment of", amount, "using UPI ID", self.upi_id)
        print("Payment Successful via UPI")


def make_payment(processor, amount):
    processor.process_payment(amount)


cc = CreditCardPayment("1234567890123456")
make_payment(cc, 500)

pp = PayPalPayment("user@example.com")
make_payment(pp, 300)

upi = UPIPayment("user@upi")
make_payment(upi, 200)
```
